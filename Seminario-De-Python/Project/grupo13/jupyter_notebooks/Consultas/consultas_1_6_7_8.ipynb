7. Listar las 5 Jurisdicciones con mayor porcentaje de población en situación de calle.
Mostrar el nombre de la jurisdicción y el porcentaje.

```python
from collections import Counter
        
data = 'Porcentaje total en situación de calle'
jurisdiction = {}

with C2022_DATA_MODIFIED.open(mode='r', encoding='utf-8') as csv_c2022:
    reader = csv.DictReader(csv_c2022)
    next(reader) # la primer línea es todo el país

    for row in reader:
        if len(jurisdiction) <= 4:
            jurisdiction[row['Jurisdicción']] = row[data]

        else:
            # Si la lista esta "llena" entonces me fijo si cambiar algún diccionario dentro de la lista
            dict_counter = Counter(jurisdiction)
            name = dict_counter.most_common()[-1][0]
            if jurisdiction[name] < row[data]:
                #change the key and value for the key and value from row
                del jurisdiction[name]
                jurisdiction[row['Jurisdicción']] = row[data]

print(jurisdiction)          
```

8. Obtener la jurisdicción donde la brecha entre las poblaciones diferenciadas según
sexo registrado al nacer es mayor. Informar nombra de Jurisdicción y cantidad de la brecha poblacional.

```python
def isnegative(n):
    return n < 0

man = 'Varones Total de población'
women = 'Mujeres Total de población'

with C2022_DATA_MODIFIED.open(mode='r', encoding='utf-8') as csv_c2022:
    reader = csv.DictReader(csv_c2022)
    next(reader)
    max_jurisdiction = None
    max_diff = -1
    for row in reader:
        diff = abs(int(row[man]) - int(row[women]))
        if diff > max_diff:
            max_jurisdiction = row['Jurisdicción']
            max_diff = diff

print(f"Jurisdicción: {max_jurisdiction}, con una brecha de {max_diff}")
```
