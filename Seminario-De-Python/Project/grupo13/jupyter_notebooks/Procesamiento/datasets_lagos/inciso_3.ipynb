## Importo la libreria csv y las dos direcciones

```python
import csv
import sys
from pathlib import Path
sys.path.append(str(Path('..').resolve().parent.parent))
from modules.paths import AR_LAKES, AR_LAKES_MODIFIED

import modulo as lagos_fun
```

# Abro los archivos como lectura y escritura para no modificar los datasets originales

```python
def process_lakes_dataset():
    try:
        with (
            AR_LAKES.open(mode = 'r', encoding='utf-8') as csv_lakes,
            AR_LAKES_MODIFIED.open(mode = 'w', encoding='utf-8', newline='') as csv_lakes_modified
            ):
            reader = csv.DictReader(csv_lakes)

            # Actualizo el nombre de las columnas, dejandolo en orden
            keyvalues = ['Latitud' if keyvalue == "Coordenadas" else keyvalue for keyvalue in reader.fieldnames]
            keyvalues += ["Longitud", "Sup Tamaño"]

            # Comienzo con los valores llave del diccionario
            writer = csv.DictWriter(csv_lakes_modified, fieldnames=keyvalues)
            # Escribo los valores llave
            writer.writeheader()
            
            # Recorro el csv reader
            for row in reader:
                # new_row es una linea con latitud y longitud insertadas
                new_row = lagos_fun.change_row(row)
                # Escribe en el diccionario el valor del resultado de la funcion de la linea no modificada
                new_row["Sup Tamaño"] = lagos_fun.surface_size(float(row['Superficie (km²)']))
                writer.writerow(new_row)
    except FileNotFoundError:
        print('Error, el archivo AR_LAKES no fue encontrado')

    
```

#### Invoco la función encargada de procesar los lagos


```python
process_lakes_dataset()
```
