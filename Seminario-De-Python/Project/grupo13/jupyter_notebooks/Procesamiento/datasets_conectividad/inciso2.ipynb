### **Reemplazo en el Dataset (B)**

Importamos las libreria csv, sys y las direcciones de los archivos.

```python
import csv
import sys
from pathlib import Path
sys.path.append(str(Path('..').resolve().parent.parent))
from modules.paths import CONNECTIONS_DATA, CONNECTIONS_DATA_MODIFIED
import modulo as conectivity_functions
```

Defino una funcion encargada de recorrer el reader e invocar a las funciones para procesar el dataset:

```python
def modified_new_dataset(reader, writer):

    for line in reader:
        
        has_connectivity = conectivity_functions.review(line)

        conectivity_functions.writing_has_conectivity(has_connectivity, line)

        writer.writerow(line)
```

Abro el archivo original como lectura y otro archivo como escritura:

```python
def process_connection_dataset():
    try:
        with(CONNECTIONS_DATA.open(mode='r', newline='', encoding='UTF-8') as file,
            CONNECTIONS_DATA_MODIFIED.open(mode='w', newline='', encoding='UTF-8') as write_file
        ):

                    reader = csv.DictReader(file) 
                    header = [key for key in reader.fieldnames] #obtengo las claves 
                    header.append('Posee_Conectividad') #agrego una nueva celda al encabezado.
                    writer = csv.DictWriter(write_file, fieldnames=header)
                    writer.writeheader() #escribo las claves

                    modified_new_dataset(reader, writer)
    except FileNotFoundError:
           print('Error, el archivo CONNECTIONS_DATA no existe')
```

Invocamos la funcion:

```python
process_connection_dataset()
```
