# Importo los paths y la librería csv

```python
import csv
import sys
from pathlib import Path
sys.path.append(str(Path('..').resolve().parent.parent))
from modules.paths import C2022_DATA, C2022_DATA_MODIFIED
```

#### Abro los archivos
##### C2022_DATA se abre en "r", sólo lectura, para no modificar los datos originales.
##### C2022_DATA_MODIFIED se abre modo "w", lo crea si no existe lo pisa caso contrario

#### Copio los campos del archivo lectura y lo escribo en el de escritura, luego, reemplazo los valores "///" y "-" por 0, luego modifico el valor con la cuenta, población calle sobre total y le agrego el porcentaje


```python
def process_census_dataset():
    try:
        with (
            C2022_DATA.open(mode = 'r', encoding='utf-8') as csv_c2022,
            C2022_DATA_MODIFIED.open(mode='w', encoding='utf-8', newline='') as csv_c2022_modified
            ):
            reader = csv.DictReader(csv_c2022)
            keyvalues = reader.fieldnames + ["Porcentaje total en situación de calle"]

            # instancio el escritor
            writer = csv.DictWriter(csv_c2022_modified, fieldnames=keyvalues)
            # valores llave
            writer.writeheader()

            # Recorro el reader
            for row in reader: 
                
                # Reemplazo los valores por 0 antes de utilizarlos para la cuenta 
                for key, value in row.items():
                    if (value == '///' or value == '-'):
                        row[key] = '0'
                # Hago la cuenta, lo asigno y lo escribo
                row["Porcentaje total en situación de calle"] = str(round((float(row["Población en situación de calle(²)"]) * 100) / float(row['Total de población']), 3)) + "%"
                writer.writerow(row)
    except FileNotFoundError:
            print('Error, el archivo C2022_DATA no fue encontrado')
```

```python
process_census_dataset()
```
